รันเซลล์ด้านล่างตามลำดับ (เมนู **Runtime → Run all**) แล้วคลิกลิงก์ที่ขึ้นมาเพื่อเปิดแอปอ่านนิยาย

- แอปนี้เป็นหน้าเว็บล้วน (HTML/JS) Colab ทำหน้าที่แค่เปิดเซิร์ฟเวอร์ให้
- เสียงอ่าน (TTS) ใช้เสียงของเบราว์เซอร์คุณเอง จึงทำงานได้ตามปกติ
- ถ้าปิด/รีสตาร์ท Runtime ลิงก์จะใช้ไม่ได้ ต้องรันใหม่

```python
# 1) ดึงโค้ดแอปจาก GitHub
REPO_URL = "https://github.com/OniSo33/Novell"
BRANCH = "main"  #@param {type:"string"}

import os, subprocess
APP_DIR = "/content/Novell"
if os.path.isdir(APP_DIR):
    subprocess.run(["git", "-C", APP_DIR, "fetch", "origin", BRANCH], check=True)
    subprocess.run(["git", "-C", APP_DIR, "checkout", "-B", BRANCH, "FETCH_HEAD"], check=True)
else:
    subprocess.run(["git", "clone", "--depth", "1", "-b", BRANCH, REPO_URL, APP_DIR], check=True)
print(sorted(os.listdir(APP_DIR)))
```

```python
# 2) เปิดเว็บเซิร์ฟเวอร์ (ทำงานเบื้องหลัง)
import subprocess, time, socket
PORT = 8000

def port_open(p):
    with socket.socket() as s:
        return s.connect_ex(("127.0.0.1", p)) == 0

if not port_open(PORT):
    server = subprocess.Popen(
        ["python3", "-m", "http.server", str(PORT), "--bind", "127.0.0.1"],
        cwd=APP_DIR, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    for _ in range(50):
        if port_open(PORT):
            break
        time.sleep(0.1)
print("Server running:", port_open(PORT))
```

```python
# 3) เปิดแอป — คลิกลิงก์ที่ขึ้นมา (เปิดเป็นแท็บใหม่ ใช้งานได้ครบที่สุด)
from google.colab import output
output.serve_kernel_port_as_window(PORT, path="/index.html", anchor_text="📖 เปิด OniSoReader")
```
